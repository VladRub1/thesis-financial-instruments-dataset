Plot summary figures from the latest run outputs.

```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Updated annual enterprise cost figures (USD)
# Consistent with the revised table:
# OCR Inference = 5,000
# Local LLM Inference = 10,000
# Rule Validation / Post-Processing = 1,000
# Data Storage = 2,000
# Logging & Audit = 1,000
# Hosting / Service Operation = 1,000
# Annotation (future extension) = 5,000
# DevOps & IT Operations = 10,000
# Team Salaries = 120,000
# Total = 155,000
# -----------------------------

components = {
    "Personnel\n(Engineering + Ops)": 120_000,
    "Compute\n(OCR + local LLM)": 15_000,
    "Operations\n& Support": 10_000,
    "Storage + Audit\n+ App overhead": 5_000,
    "Annotation\n(future extension)": 5_000,
}

labels = list(components.keys())
values = np.array(list(components.values()), dtype=float)
total = values.sum()
percentages = values / total * 100

# -----------------------------
# Figure setup
# -----------------------------
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.titlesize": 16,
    "figure.dpi": 180,
})

fig, ax = plt.subplots(figsize=(8.2, 6.2), facecolor="white")

# Use grayscale tones for a journal-like aesthetic
colors = ["0.25", "0.45", "0.60", "0.75", "0.88"]

wedges, _ = ax.pie(
    values,
    startangle=90,
    counterclock=False,
    colors=colors,
    wedgeprops=dict(width=0.42, edgecolor="white", linewidth=1.2),
)

# -----------------------------
# Percentage labels inside slices
# Only show for slices >= 3% to avoid clutter
# -----------------------------
for w, pct in zip(wedges, percentages):
    if pct < 3:
        continue
    ang = (w.theta2 + w.theta1) / 2.0
    x = 0.53 * np.cos(np.deg2rad(ang))
    y = 0.53 * np.sin(np.deg2rad(ang))
    ax.text(
        x, y, f"{pct:.1f}%",
        ha="center", va="center",
        fontsize=11, color="white" if pct > 20 else "black"
    )

# -----------------------------
# External labels with value + share
# -----------------------------
for w, label, val, pct in zip(wedges, labels, values, percentages):
    ang = (w.theta2 + w.theta1) / 2.0
    x = np.cos(np.deg2rad(ang))
    y = np.sin(np.deg2rad(ang))

    # label anchor
    rx, ry = 1.18 * x, 1.18 * y
    ha = "left" if x >= 0 else "right"

    ax.annotate(
        f"{label}\n${val:,.0f} ({pct:.1f}%)",
        xy=(0.82 * x, 0.82 * y),
        xytext=(rx, ry),
        ha=ha,
        va="center",
        fontsize=10.5,
        arrowprops=dict(
            arrowstyle="-",
            lw=0.8,
            color="0.35",
            shrinkA=0,
            shrinkB=0,
            connectionstyle="arc3,rad=0.08"
        ),
    )

# Center text
ax.text(0, 0.06, "Annual\nEnterprise Cost", ha="center", va="center",
        fontsize=13, fontweight="semibold")
ax.text(0, -0.11, f"${total:,.0f}", ha="center", va="center",
        fontsize=15)

ax.set_title("Enterprise Deployment Cost Split", pad=20)
ax.set(aspect="equal")

# Add a clean caption-like footnote
fig.text(
    0.5, 0.03,
    "Illustrative annual cost structure for enterprise deployment. "
    "Annotation is treated as a future extension rather than a core runtime requirement.",
    ha="center", va="center", fontsize=9.5
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

# Save in high quality for thesis / article use
plt.savefig("cost_split_diagram_v2.png", bbox_inches="tight", dpi=300)
plt.savefig("cost_split_diagram_v2.pdf", bbox_inches="tight")
plt.show()
```
